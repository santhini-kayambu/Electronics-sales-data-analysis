### Data Visualization

Data Visualization is the process of presenting data through graphical representations such as bar charts, pie charts, line charts, scatter plots, and heatmaps.

In this step, we visualize electronics sales data to understand product performance, brand performance, regional sales, quantity distribution, and sales trends.

These visualizations help identify important patterns and support better business decision-making.

```python
plt.figure(figsize=(8,5))

product_sales = df.groupby('Product')['sales value'].sum()

sns.barplot(x=product_sales.index, y=product_sales.values)

plt.title("Product-wise Sales Analysis")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(12,6))

brand_sales = df.groupby('Brand')['sales value'].sum().nlargest(10)

sns.barplot(x=brand_sales.values, y=brand_sales.index)

plt.title("Top 10 Brands by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Brand")

plt.show()
```

```python
plt.figure(figsize=(9,5))

region_sales = df.groupby('Region')['sales value'].sum().sort_values(ascending=False)

sns.barplot(x=region_sales.index, y=region_sales.values)

plt.title("Region-wise Sales Performance")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
product_sales = df.groupby('Product')['sales value'].sum()

plt.figure(figsize=(7,7))

plt.pie(
    product_sales,
    labels=product_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Product-wise Sales Distribution")
plt.show()
```

```python
monthly_sales = df.groupby(
    df['Inward Date'].dt.to_period('M')
)['sales value'].sum()

plt.figure(figsize=(14,6))

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
yearly_sales = df.groupby(
    df['Inward Date'].dt.year
)['sales value'].sum()

plt.figure(figsize=(8,5))

yearly_sales.plot(kind='bar')

plt.title("Yearly Sales Comparison")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.histplot(df['Price'], bins=30, kde=True)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x='Price',
    y='sales value',
    alpha=0.5
)

plt.title("Price vs Sales Value")
plt.xlabel("Price")
plt.ylabel("Sales Value")

plt.show()
```

```python
plt.figure(figsize=(8,5))

correlation = df[
    ['Price', 'Quantity Sold', 'sales value']
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Sales Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Product',
    y='Quantity Sold'
)

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.show()
```

### Evidence-Based Insights

```python
# EVIDENCE-BASED INSIGHTS

print("=" * 50)
print("       EVIDENCE-BASED BUSINESS INSIGHTS")
print("=" * 50)

# 1. Overall Sales Performance
print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 40)

print("Total Sales:", df['sales value'].sum())
print("Total Quantity Sold:", df['Quantity Sold'].sum())
print("Average Sales:", round(df['sales value'].mean(), 2))
print("Total Products:", df['Product'].nunique())
print("Total Brands:", df['Brand'].nunique())
print("Total Regions:", df['Region'].nunique())

# 2. Brand-wise Sales Evidence
print("\n2. BRAND-WISE SALES ANALYSIS")
print("-" * 40)

brand_sales = df.groupby('Brand')['sales value'].sum().sort_values(ascending=False)
print("Top 5 Brands:\n", brand_sales.head(5))
print("Highest Sales Brand:", brand_sales.idxmax())

# 3. Product-wise Sales Evidence
print("\n3. PRODUCT-WISE SALES ANALYSIS")
print("-" * 40)

product_sales = df.groupby('Product')['sales value'].sum().sort_values(ascending=False)
print(product_sales)
print("Best Performing Product:", product_sales.idxmax())

# 4. Region-wise Sales Evidence
print("\n4. REGION-WISE SALES ANALYSIS")
print("-" * 40)

region_sales = df.groupby('Region')['sales value'].sum().sort_values(ascending=False)
print(region_sales)
print("Highest Sales Region:", region_sales.idxmax())

# 5. Monthly and Yearly Sales Evidence
print("\n5. TIME SERIES ANALYSIS")
print("-" * 40)

df['Inward Date'] = pd.to_datetime(df['Inward Date'])

monthly_sales = df.groupby(
    df['Inward Date'].dt.to_period('M')
)['sales value'].sum()

yearly_sales = df.groupby(
    df['Inward Date'].dt.year
)['sales value'].sum()

print("Monthly Sales:\n", monthly_sales)
print("\nYearly Sales:\n", yearly_sales)

# 6. Correlation Evidence
print("\n6. CORRELATION ANALYSIS")
print("-" * 40)

print(df[['Price', 'Quantity Sold', 'sales value']].corr())

print("\n" + "=" * 50)
print("EVIDENCE-BASED INSIGHTS GENERATED SUCCESSFULLY!")
print("=" * 50)
```
